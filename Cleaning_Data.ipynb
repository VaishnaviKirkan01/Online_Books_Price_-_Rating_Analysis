```python
# changing names for easier understanding

df.columns=df.columns.str.strip().str.replace({
    'price_tag':'price',
    'availible_tag':'avalibility',
    'category_tag':'category',
    'rating_tag':'rating'
})
```

```python
# replacing the categorical data into numerical

df['rating']=df['rating'].str.replace({
    "One":"1",
    "Two":'2',
    "Three":'3',
    "Four":'4',
    "Five":'5'
})
```

```python
# converting the rating into numrerical 

df['rating']=pd.to_numeric(df['rating'],errors="coerce")
```

```python
# standardise the category column

df['category']=df['category'].str.strip().str.title()
```
